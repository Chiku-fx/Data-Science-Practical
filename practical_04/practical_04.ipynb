# Practical 4: Data visualisation

## Aim
Unique instructors, bar chart by division and scatter plots of age against evaluation.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("TeachingRatings.csv")
```

```python
unique = df.drop_duplicates("prof")
print("Total observations:", len(df))
print("Unique instructors:", len(unique))
print("\nAge - all observations: mean %.2f, std %.2f" % (df["age"].mean(), df["age"].std()))
print("Age - unique instructors: mean %.2f, std %.2f" % (unique["age"].mean(), unique["age"].std()))
```

```python
division_avg = df.groupby("division")["eval"].mean()
print("\nAverage evaluation by division:")
print(division_avg.round(4))
division_avg.plot(kind="bar", color=["teal", "salmon"], rot=0)
plt.title("Average Evaluation by Division")
plt.ylabel("Average evaluation")
plt.show()
```

```python
plt.scatter(df["age"], df["eval"], alpha=0.6)
plt.title("Age vs Evaluation")
plt.xlabel("Age")
plt.ylabel("Evaluation")
plt.show()
```

```python
for name, part in df.groupby("gender"):
    plt.scatter(part["age"], part["eval"], alpha=0.6, label=name)
plt.legend()
plt.title("Age vs Evaluation by Gender")
plt.xlabel("Age")
plt.ylabel("Evaluation")
plt.show()
```

```python
for (g, t), part in df.groupby(["gender", "tenure"]):
    plt.scatter(part["age"], part["eval"], alpha=0.6, label=g + " / tenure " + t)
plt.legend()
plt.title("Age vs Evaluation by Gender and Tenure")
plt.xlabel("Age")
plt.ylabel("Evaluation")
plt.show()
```

```python
print("\nCorrelation of age and evaluation: %.4f" % df["age"].corr(df["eval"]))
print("\nResult: there are 94 unique instructors in 463 rows, lower-division courses score")
print("slightly higher, and age has almost no relationship with evaluation score.")
```
