# Practical 1: NumPy and Pandas basics

## Aim
Load the Students Performance data, convert it to arrays, check shape/size/dimensions, and apply matrix operations.

```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("StudentsPerformance.csv")
```

```python
print("Rows and columns:", data.shape)
print(data.dtypes)
```

```python
scores = data[["math score", "reading score", "writing score"]].values
print("\nShape:", scores.shape)
print("Size:", scores.size)
print("Dimensions:", scores.ndim)
```

```python
part = data[["math score", "reading score"]].iloc[:4]
print("\nSelected rows:")
print(part)
```

```python
block = part.values
first, second = block[:2], block[2:]
print("\nFirst matrix:")
print(first)
print("Second matrix:")
print(second)
```

```python
print("\nAddition:")
print(np.add(first, second))
print("Subtraction:")
print(np.subtract(first, second))
print("Element-wise multiplication:")
print(np.multiply(first, second))
print("Matrix product:")
print(first @ second)
print("Inverse of first matrix:")
print(np.round(np.linalg.inv(first), 4))
print("Transpose of first matrix:")
print(first.transpose())
```

```python
encoded = pd.get_dummies(data, dtype=int)
print("\nShape after one-hot encoding:", encoded.shape)
print(encoded.iloc[:3, :8])
```

```python
print("\nResult: the data was converted to arrays and all matrix operations ran successfully.")
```
