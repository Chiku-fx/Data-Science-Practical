# Practical 5: Probability and z-test

## Aim
Normal probabilities for evaluation scores and a two-tailed z-test.

```python
from math import sqrt
import pandas as pd
from scipy.stats import norm
```

```python
df = pd.read_csv("TeachingRatings.csv")
```

```python
m = df["eval"].mean()
sd = df["eval"].std()
print("Mean evaluation: %.4f" % m)
print("Standard deviation: %.4f" % sd)
```

```python
z = (4.5 - m) / sd
print("\nP(score > 4.5)")
print("z = %.4f" % z)
print("Normal model: %.4f (%.2f%%)" % (norm.sf(z), norm.sf(z) * 100))
actual = (df["eval"] > 4.5).mean()
print("From data:    %.4f (%.2f%%)" % (actual, actual * 100))
```

```python
lo, hi = (3.5 - m) / sd, (4.2 - m) / sd
p = norm.cdf(hi) - norm.cdf(lo)
print("\nP(3.5 < score < 4.2)")
print("z values = %.4f, %.4f" % (lo, hi))
print("Normal model: %.4f (%.2f%%)" % (p, p * 100))
actual = df["eval"].between(3.5, 4.2, inclusive="neither").mean()
print("From data:    %.4f (%.2f%%)" % (actual, actual * 100))
```

```python
mu, sigma, n, xbar = 12, 5.5, 36, 10.7
se = sigma / sqrt(n)
zt = (xbar - mu) / se
pv = 2 * norm.sf(abs(zt))
crit = norm.ppf(0.975)
print("\nTwo-tailed z-test (H0: mean = 12, alpha = 0.05)")
print("Standard error: %.4f" % se)
print("z statistic: %.4f" % zt)
print("p-value: %.4f" % pv)
print("Critical value: +/- %.4f" % crit)
print("Decision:", "Reject H0" if abs(zt) > crit else "Fail to reject H0")
```

```python
print("\nResult: P(score > 4.5) is about 18.29%, P(3.5 < score < 4.2) is about 45.73%,")
print("and the z-test does not reject H0, so no significant difference in means is found.")
```
